# Reactor-Particle Coupling with Maxwell-Stefan Film Transfer

This exercise is based on the lecture notebook `L8_reactor_particle_maxwell_stefan_coupling.ipynb`.

## Task

Extend reactor-particle coupling from a scalar film-transfer coefficient to a multicomponent Maxwell-Stefan film model.

## Questions

1. Reproduce the base-case coupled reactor-particle calculation.
2. Plot reactor bulk concentrations, particle boundary concentrations, and particle-center concentrations along the reactor.
3. Compute the outlet conversion of A and the minimum concentration in the full state vector.
4. Compare the film-transfer residual and the particle residual at convergence.
5. Change one binary Maxwell-Stefan diffusivity and describe how the outlet conversion and boundary concentration of A respond.

## Deliverables

Submit plots of the axial profiles, residual checks, and a short explanation of why multicomponent film transfer is not equivalent to independent Fickian film coefficients.
