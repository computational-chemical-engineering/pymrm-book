# Monolithic 2D Membrane Module

This exercise is based on the lecture notebook `L8_simple_monolithic_membrane_module.ipynb`.

## Task

Use the monolithic 2D membrane module to study counter-current retentate/permeate coupling.

## Questions

1. Reproduce the base-case retentate and permeate concentration fields.
2. Compute cup-mixing average concentrations on both sides of the membrane.
3. Verify that the total retentate-plus-permeate molar flow is conserved.
4. Sweep the membrane permeability over at least four values and plot retentate outlet concentration versus permeability.
5. Explain why the interface must be coupled implicitly for large permeability.

## Deliverables

Submit the concentration-field plot, membrane-flux plot, permeability sweep, and conservation check.
